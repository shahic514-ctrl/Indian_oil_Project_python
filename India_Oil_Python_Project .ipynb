# Data Cleaning

## Check missing values

```python
df.isnull().sum()
```

## Check duplicate records

```python
df.duplicated().sum()
```

## Remove duplicates

```python
df = df.drop_duplicates()
```

# Convert Date column

```python
df['Date'] = pd.to_datetime(
    df['Date'],
    dayfirst=True
)
```

 # Sort data by date

```python
df = df.sort_values('Date')

df.head()
```

## Check data types

```python
df.dtypes
```

 # Basic Analysis

 ## Highest Closing Price

```python
highest_close = df['Close'].max()

print("Highest Closing Price:", highest_close)
```

## Lowest Closing Price

```python
lowest_close = df['Close'].min()

print("Lowest Closing Price:", lowest_close)
```

 ## Average Closing Price

```python
average_close = df['Close'].mean()

print("Average Closing Price:", average_close)
```

 # Highest Trading Volume

```python
highest_volume = df['Volume'].max()

print("Highest Trading Volume:", highest_volume)
```

 ## Create New Columns

# Daily Return

```python
df['Daily_Return'] = df['Close'].pct_change() * 100
```

 ## Price Change

```python
df['Price_Change'] = df['Close'] - df['Open']
```

# Daily Volatility

```python
df['Volatility'] = df['High'] - df['Low']
```

# Closing Price Trend

```python
plt.figure(figsize=(14, 6))

plt.plot(
    df['Date'],
    df['Close']
)

plt.title('Indian Oil Closing Price Trend')
plt.xlabel('Date')
plt.ylabel('Closing Price')

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

# Open vs Close Price 

```python
plt.figure(figsize=(14, 6))

plt.plot(
    df['Date'],
    df['Open'],
    label='Open'
)

plt.plot(
    df['Date'],
    df['Close'],
    label='Close'
)

plt.title('Indian Oil Open vs Close Price')
plt.xlabel('Date')
plt.ylabel('Price')

plt.legend()
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

# High vs Low Price 

```python
plt.figure(figsize=(14, 6))

plt.plot(
    df['Date'],
    df['High'],
    label='High'
)

plt.plot(
    df['Date'],
    df['Low'],
    label='Low'
)

plt.title('Indian Oil High vs Low Price')
plt.xlabel('Date')
plt.ylabel('Price')

plt.legend()
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

# Trading Volume

```python
plt.figure(figsize=(14, 6))

plt.plot(
    df['Date'],
    df['Volume']
)

plt.title('Indian Oil Trading Volume')
plt.xlabel('Date')
plt.ylabel('Volume')

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

# Year-wise Average Closing Price

```python
yearly_close = (
    df.groupby('Year')['Close']
    .mean()
    .reset_index()
)

yearly_close
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=yearly_close,
    x='Year',
    y='Close'
)

plt.title('Year-wise Average Closing Price')
plt.xlabel('Year')
plt.ylabel('Average Closing Price')

plt.show()
```

# Monthly Average Closing Price

```python
monthly_close = (
    df.groupby('Month')['Close']
    .mean()
    .reset_index()
)

monthly_close
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=monthly_close,
    x='Month',
    y='Close'
)

plt.title('Monthly Average Closing Price')
plt.xlabel('Month')
plt.ylabel('Average Closing Price')

plt.xticks(rotation=45)

plt.show()
```

# Moving Average

```python
df['MA_30'] = df['Close'].rolling(30).mean()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    df['Date'],
    df['Close'],
    label='Closing Price'
)

plt.plot(
    df['Date'],
    df['MA_30'],
    label='30 Day Moving Average'
)

plt.title('Indian Oil Closing Price with 30-Day Moving Average')

plt.xlabel('Date')
plt.ylabel('Price')

plt.legend()
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

# Daily Return Distribution

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df['Daily_Return'].dropna(),
    bins=30,
    kde=True
)

plt.title('Indian Oil Daily Return Distribution')
plt.xlabel('Daily Return (%)')
plt.ylabel('Frequency')

plt.show()
```

# Correlation Analysis

```python
correlation = df[
    ['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']
].corr()

correlation
```

# Heatmap

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Indian Oil Stock Data Correlation')

plt.show()
```
